## 📚 Teoría: Trabajar con JSON en Python

### ¿Qué es JSON?
**JSON** (JavaScript Object Notation) es un formato de texto para intercambiar datos entre sistemas. Es ligero y fácil de leer tanto para humanos como para máquinas.

### 🔄 Dos Operaciones Principales:

1. **Serialización** (Python → JSON): Convertir objetos Python a texto JSON
2. **Deserialización** (JSON → Python): Convertir texto JSON a objetos Python

---

### 📊 Equivalencias Python ↔ JSON:

| Python | JSON |
|--------|------|
| `dict` | `object` |
| `list`, `tuple` | `array` |
| `str` | `string` |
| `int`, `float` | `number` |
| `True` | `true` |
| `False` | `false` |
| `None` | `null` |

### 🔧 Métodos Principales del Módulo `json`

#### 1. **`json.dumps()`** - Python a JSON (String)
Convierte un objeto Python en una cadena JSON.

**Sintaxis:** `json.dumps(obj, indent=None, sort_keys=False, ensure_ascii=True)`

**Parámetros importantes:**
- `obj`: El objeto Python a convertir
- `indent`: Número de espacios para indentar (hace el JSON legible)
- `sort_keys`: Si es `True`, ordena las claves alfabéticamente
- `ensure_ascii`: Si es `False`, permite caracteres especiales (ñ, á, etc.)

#### 2. **`json.dump()`** - Python a JSON (Archivo)
Igual que `dumps()`, pero guarda directamente en un archivo.

**Sintaxis:** `json.dump(obj, file, indent=None, sort_keys=False)`

#### 3. **`json.loads()`** - JSON (String) a Python
Convierte una cadena JSON en un objeto Python.

**Sintaxis:** `json.loads(json_string)`

#### 4. **`json.load()`** - JSON (Archivo) a Python
Lee un archivo JSON y lo convierte a objeto Python.

**Sintaxis:** `json.load(file)`

### 🌐 Métodos de la Librería `requests`

Cuando trabajas con APIs, usas la librería `requests` para hacer peticiones HTTP.

#### **Métodos HTTP más comunes:**

| Método | Propósito | Uso |
|--------|-----------|-----|
| **GET** | Obtener datos | `requests.get(url)` |
| **POST** | Enviar datos nuevos | `requests.post(url, json=data)` |
| **PUT** | Actualizar datos existentes | `requests.put(url, json=data)` |
| **DELETE** | Eliminar datos | `requests.delete(url)` |
| **PATCH** | Actualizar parcialmente | `requests.patch(url, json=data)` |

---

#### **`requests.get()`** - El más usado
Obtiene datos de una API.

**Sintaxis:** `requests.get(url, params=None, headers=None, timeout=None)`

**Parámetros importantes:**
- `url`: La dirección de la API
- `params`: Diccionario con parámetros de la URL (?clave=valor)
- `headers`: Diccionario con encabezados HTTP (autenticación, tipo de contenido, etc.)
- `timeout`: Segundos máximos de espera

**Métodos del objeto `response`:**
- `.json()`: Convierte la respuesta JSON a diccionario Python
- `.text`: Obtiene la respuesta como texto plano
- `.status_code`: Código HTTP (200=éxito, 404=no encontrado, etc.)
- `.headers`: Encabezados de la respuesta
- `.content`: Contenido en bytes (útil para archivos)

## 🎨 Ejemplos Prácticos: Imprimir JSON de Forma Legible

Vamos a ver diferentes formas de imprimir tu respuesta JSON de manera bonita y fácil de leer.

````python
# Imports necesarios para los ejemplos
import json
import requests
from datetime import datetime

# Ejemplo de respuesta JSON (simulado para la demostración)
response = {
    "location": {
        "name": "Madrid",
        "country": "Spain"
    },
    "current": {
        "temp_c": 8.2,
        "humidity": 65,
        "wind_kph": 15.5,
        "condition": {
            "text": "Overcast"
        }
    }
}
````

````python
# ❌ FORMA FEA: Imprimir directamente (difícil de leer)
# Si haces esto, todo sale en una línea muy larga
print("--- FORMA FEA (NO HAGAS ESTO) ---")
print(response)  # Todo en una línea, difícil de leer
````

````python
# ✅ FORMA BONITA 1: Usar json.dumps() con indent
# indent=4 significa que cada nivel se indenta con 4 espacios
# Esto hace que el JSON sea MUY legible
print("\n--- FORMA BONITA 1: json.dumps() con indent ---")
print(json.dumps(response, indent=4))  # ← Cada nivel indentado con 4 espacios
````

````python
# ✅ FORMA BONITA 2: Con indent Y sort_keys
# sort_keys=True ordena las claves alfabéticamente
# ensure_ascii=False permite mostrar caracteres especiales (ñ, á, ü, etc.)
print("\n--- FORMA BONITA 2: Con claves ordenadas y caracteres especiales ---")
print(json.dumps(response, indent=4, sort_keys=True, ensure_ascii=False))
````

````python
# ✅ FORMA BONITA 3: Usar pprint (Pretty Print)
# pprint es una librería de Python específica para imprimir estructuras de datos
# Funciona bien con diccionarios y listas anidadas
from pprint import pprint  # Importar pretty print

print("\n--- FORMA BONITA 3: Usando pprint ---")
pprint(response, indent=4, width=80)  # width=80 limita el ancho de línea
````

### 🎯 Acceder a Datos Específicos del JSON

Cuando tienes un JSON, puedes navegar por él como si fuera un diccionario Python.

````python
# Ejemplo: Extraer información específica del clima

# 1. Ver todas las claves principales del JSON
print("Claves principales del JSON:")
print(response.keys())

print("\n" + "="*50 + "\n")

# 2. Acceder a datos anidados usando corchetes []
# El JSON tiene esta estructura: response['location']['name']
ciudad = response['location']['name']  # Acceder a location > name
pais = response['location']['country']  # Acceder a location > country
temp_actual = response['current']['temp_c']  # Acceder a current > temp_c
condicion = response['current']['condition']['text']  # Acceder a current > condition > text

# 3. Imprimir los datos extraídos de forma legible
print(f"📍 Ciudad: {ciudad}, {pais}")
print(f"🌡️  Temperatura actual: {temp_actual}°C")
print(f"☁️  Condición: {condicion}")
````

````python
# ⚠️ ACCESO SEGURO: Usar .get() para evitar errores

# Problema con corchetes []: Si la clave no existe, Python arroja un error (KeyError)
# Solución: Usar .get() que devuelve None (o un valor por defecto) si no existe

# ❌ FORMA INSEGURA (puede dar error si la clave no existe)
# temperatura = response['datos']['temp']  # ← Error si 'datos' no existe

# ✅ FORMA SEGURA: Usar .get() con valor por defecto
temperatura = response.get('current', {}).get('temp_c', 'No disponible')
humedad = response.get('current', {}).get('humidity', 'No disponible')
viento = response.get('current', {}).get('wind_kph', 'No disponible')

# Imprimir los datos de forma segura
print("\n📊 Datos meteorológicos:")
print(f"  Temperatura: {temperatura}°C")
print(f"  Humedad: {humedad}%")
print(f"  Viento: {viento} km/h")

# Explicación de .get():
# response.get('clave', valor_por_defecto)
# - Si 'clave' existe, devuelve su valor
# - Si 'clave' NO existe, devuelve 'valor_por_defecto' (sin error)
````

### 💾 Guardar y Cargar JSON desde Archivos

Es común guardar respuestas de APIs en archivos JSON para no hacer peticiones repetidas.

````python
# 💾 GUARDAR JSON en un archivo

# Crear un nombre de archivo con la fecha y hora actual
nombre_archivo = f"clima_ejemplo_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"

# Abrir el archivo en modo escritura ('w' = write)
# 'with' asegura que el archivo se cierre automáticamente
with open(nombre_archivo, 'w', encoding='utf-8') as archivo:
    # json.dump() guarda el diccionario Python como JSON en el archivo
    # indent=4 hace que el archivo sea legible
    # ensure_ascii=False permite caracteres especiales (ñ, á, etc.)
    json.dump(response, archivo, indent=4, ensure_ascii=False)

print(f"✅ JSON guardado en: {nombre_archivo}")
````

````python
# 📂 CARGAR JSON desde un archivo

# Abrir el archivo en modo lectura ('r' = read)
with open(nombre_archivo, 'r', encoding='utf-8') as archivo:
    # json.load() lee el archivo JSON y lo convierte a diccionario Python
    data_cargada = json.load(archivo)

# Verificar que se cargó correctamente
print(f"✅ JSON cargado desde: {nombre_archivo}")
print(f"📊 Tipo de dato: {type(data_cargada)}")  # Debe ser <class 'dict'>
print(f"🔑 Claves principales: {list(data_cargada.keys())}")

# Ahora puedes usar data_cargada igual que response
print(f"\nTemperatura del archivo: {data_cargada['current']['temp_c']}°C")
````

### 🔄 Trabajar con Strings JSON (Conversión)

A veces recibes JSON como texto (string) y necesitas convertirlo a diccionario Python.

````python
# 🔄 Conversión: Diccionario ↔ String JSON

# 1. De diccionario Python a string JSON (SERIALIZACIÓN)
mi_diccionario = {
    "nombre": "María",
    "edad": 25,
    "ciudad": "Madrid",
    "hobbies": ["leer", "programar", "viajar"]
}

# Convertir a string JSON
json_string = json.dumps(mi_diccionario, indent=4, ensure_ascii=False)
print("📦 Diccionario convertido a JSON string:")
print(json_string)
print(f"Tipo: {type(json_string)}")  # <class 'str'>

print("\n" + "="*50 + "\n")

# 2. De string JSON a diccionario Python (DESERIALIZACIÓN)
json_texto = '{"nombre": "Juan", "edad": 30, "ciudad": "Barcelona"}'

# Convertir a diccionario Python
diccionario_nuevo = json.loads(json_texto)
print("📂 JSON string convertido a diccionario:")
print(diccionario_nuevo)
print(f"Tipo: {type(diccionario_nuevo)}")  # <class 'dict'>

# Ahora puedes acceder a los valores como un diccionario normal
print(f"\nNombre: {diccionario_nuevo['nombre']}")
print(f"Ciudad: {diccionario_nuevo['ciudad']}")
````

## 💡 Mejores Prácticas y Consejos

### ✅ **Hacer:**
1. **Siempre usar `indent`** cuando imprimes JSON para debugging
2. **Usar `.get()` en lugar de `[]`** para acceder a claves que podrían no existir
3. **Validar el `status_code`** de las peticiones HTTP (200 = éxito)
4. **Guardar respuestas grandes** en archivos para no repetir peticiones
5. **Usar `ensure_ascii=False`** si trabajas con caracteres especiales (español, etc.)
6. **Agregar `timeout`** a tus peticiones requests para evitar esperas infinitas

### ❌ **Evitar:**
1. **No imprimir JSON sin `indent`** - es ilegible
2. **No acceder a claves sin validar** que existen
3. **No hacer peticiones repetidas** innecesariamente a una API
4. **No guardar credenciales en el código** (usar variables de entorno)
5. **No ignorar los códigos de error** HTTP (400, 404, 500, etc.)

### 🛡️ Ejemplo Completo: Petición HTTP con Manejo de Errores

Así es como deberías hacer una petición HTTP profesional.

````python
# 🛡️ EJEMPLO PROFESIONAL: Petición HTTP con validación completa

from requests.exceptions import Timeout, ConnectionError

def obtener_clima_seguro(ciudad, api_key):
    """
    Función que obtiene el clima de forma segura con manejo de errores.
    
    Parámetros:
        ciudad (str): Nombre de la ciudad
        api_key (str): Tu API key de weatherapi.com
    
    Retorna:
        dict: Datos del clima o None si hay error
    """
    try:
        # 1. Construir la URL
        url = f'http://api.weatherapi.com/v1/forecast.json'
        
        # 2. Definir los parámetros de la petición
        params = {
            'key': api_key,
            'q': ciudad,
            'days': 1,
            'aqi': 'no',
            'alerts': 'no'
        }
        
        # 3. Hacer la petición con timeout (espera máxima 10 segundos)
        print(f"🌐 Haciendo petición para: {ciudad}...")
        response = requests.get(url, params=params, timeout=10)
        
        # 4. Verificar que la petición fue exitosa (código 200)
        if response.status_code == 200:
            # La petición fue exitosa
            data = response.json()  # Convertir respuesta a diccionario
            print(f"✅ Datos obtenidos correctamente")
            return data
        else:
            # La petición falló (código 400, 404, 500, etc.)
            print(f"❌ Error HTTP {response.status_code}: {response.text}")
            return None
            
    except Timeout:
        # La petición tardó demasiado (más de 10 segundos)
        print("⏰ Error: La petición tardó demasiado tiempo (timeout)")
        return None
        
    except ConnectionError:
        # No hay conexión a internet o el servidor no responde
        print("🔌 Error: No se pudo conectar al servidor")
        return None
        
    except json.JSONDecodeError:
        # La respuesta no es un JSON válido
        print("📄 Error: La respuesta no es un JSON válido")
        return None
        
    except Exception as e:
        # Cualquier otro error no previsto
        print(f"⚠️ Error inesperado: {str(e)}")
        return None


# Ejemplo de uso (necesitarías tu propia API key):
# clima = obtener_clima_seguro('Madrid', 'TU_API_KEY')
# 
# if clima:
#     # Si se obtuvo el clima, mostrarlo de forma bonita
#     print("\n" + "="*50)
#     print(json.dumps(clima, indent=4, ensure_ascii=False))
# else:
#     # Si hubo error, informar al usuario
#     print("No se pudo obtener el clima")
````

## 🗺️ Entender Estructuras JSON Anidadas

Cuando trabajas con APIs, el JSON puede tener muchos niveles de anidación (como cajas dentro de cajas). Aquí aprenderás a leerlo como un mapa.

### 🧩 Los Dos Tipos de Estructuras en JSON

#### 1. **Objeto** (Diccionario en Python) → `{}`
- Colección de pares **clave: valor**
- Se accede por **nombre de clave**
- En Python: `dict`

```json
{
  "nombre": "Madrid",
  "temperatura": 15.5,
  "pais": "Spain"
}
```

**Acceso en Python:**
```python
data['nombre']         # → "Madrid"
data['temperatura']    # → 15.5
```

#### 2. **Array** (Lista en Python) → `[]`
- Colección **ordenada** de elementos
- Se accede por **índice numérico** (empieza en 0)
- En Python: `list`

```json
["Madrid", "Barcelona", "Valencia"]
```

**Acceso en Python:**
```python
ciudades[0]  # → "Madrid" (primer elemento)
ciudades[1]  # → "Barcelona" (segundo elemento)
ciudades[2]  # → "Valencia" (tercer elemento)
```

### 🚶 Cómo Navegar Paso a Paso

Para acceder a datos anidados, vas bajando nivel por nivel como si bajaras escaleras.

````python
# Ejemplo de JSON anidado (similar al de Weather API)
response = {
    "location": {
        "name": "Madrid",
        "country": "Spain"
    },
    "current": {
        "temp_c": 15.5,
        "humidity": 65
    },
    "forecast": {
        "forecastday": [
            {
                "date": "2026-02-18",
                "day": {
                    "maxtemp_c": 18.0,
                    "mintemp_c": 8.0
                },
                "hour": [
                    {"time": "2026-02-18 00:00", "temp_c": 8.5, "condition": {"text": "Clear"}},
                    {"time": "2026-02-18 01:00", "temp_c": 8.2, "condition": {"text": "Clear"}},
                    {"time": "2026-02-18 02:00", "temp_c": 7.9, "condition": {"text": "Partly cloudy"}}
                ]
            }
        ]
    }
}

# 🎯 EJEMPLOS DE NAVEGACIÓN PASO A PASO

print("="*60)
print("NIVEL 1: Acceso Simple (1 nivel)")
print("="*60)

# Acceder a las claves principales del objeto raíz
print(f"Ciudad: {response['location']['name']}")
print(f"País: {response['location']['country']}")
print(f"Temperatura actual: {response['current']['temp_c']}°C")

print("\n" + "="*60)
print("NIVEL 2: Acceso a Arrays (listas)")
print("="*60)

# Acceder al primer día del pronóstico
primer_dia = response['forecast']['forecastday'][0]  # [0] = primer elemento de la lista
print(f"Fecha del pronóstico: {primer_dia['date']}")
print(f"Temperatura máxima: {primer_dia['day']['maxtemp_c']}°C")
print(f"Temperatura mínima: {primer_dia['day']['mintemp_c']}°C")

print("\n" + "="*60)
print("NIVEL 3: Acceso Profundo (múltiples niveles)")
print("="*60)

# Acceder a la segunda hora (índice 1) del primer día
hora_01 = response['forecast']['forecastday'][0]['hour'][1]
print(f"Hora: {hora_01['time']}")
print(f"Temperatura: {hora_01['temp_c']}°C")
print(f"Condición: {hora_01['condition']['text']}")

print("\n" + "="*60)
print("DESGLOSE: ¿Cómo llegamos aquí?")
print("="*60)

# Explicación paso a paso
print("response['forecast']['forecastday'][0]['hour'][1]['temp_c']")
print("   ↓")
print("1. response → Objeto raíz (todo el JSON)")
print("2. ['forecast'] → Sección 'forecast' (objeto)")
print("3. ['forecastday'] → Lista de días (array)")
print("4. [0] → Primer día (índice 0 del array)")
print("5. ['hour'] → Lista de horas de ese día (array)")
print("6. [1] → Segunda hora (índice 1 del array)")
print("7. ['temp_c'] → Temperatura en Celsius (valor)")
print(f"\nResultado: {response['forecast']['forecastday'][0]['hour'][1]['temp_c']}°C")
````

### 🔁 Recorrer Arrays (Listas) en JSON

Cuando tienes un array, puedes recorrerlo con un bucle `for` para procesar cada elemento.

````python
# 🔄 EJEMPLO 1: Recorrer todas las horas del día

print("="*60)
print("Pronóstico por hora para el día:", response['forecast']['forecastday'][0]['date'])
print("="*60)

# Obtener el array de horas del primer día
horas_del_dia = response['forecast']['forecastday'][0]['hour']

# Recorrer cada hora
for hora_info in horas_del_dia:
    # Extraer la hora del timestamp (ej: "2026-02-18 01:00" → "01:00")
    tiempo = hora_info['time'].split()[1]  # Dividir por espacio y tomar la segunda parte
    temperatura = hora_info['temp_c']
    condicion = hora_info['condition']['text']
    
    # Imprimir información de cada hora
    print(f"⏰ {tiempo} → {temperatura}°C - {condicion}")

print("\n" + "="*60)
print("📊 ESTADÍSTICAS DEL DÍA")
print("="*60)

# Calcular estadísticas usando el array
temperaturas = [h['temp_c'] for h in horas_del_dia]  # List comprehension

print(f"Temperatura más alta: {max(temperaturas)}°C")
print(f"Temperatura más baja: {min(temperaturas)}°C")
print(f"Temperatura promedio: {sum(temperaturas) / len(temperaturas):.1f}°C")
print(f"Total de horas en el pronóstico: {len(horas_del_dia)}")
````

### 💡 Trucos y Consejos para JSON Anidado

#### ✅ **Buenas Prácticas:**

1. **Explora primero con `.keys()`**
   ```python
   print(response.keys())  # Ver las claves principales
   ```

### 🎯 Ejemplo Práctico: Extraer Datos Útiles

Imagina que quieres crear un resumen del clima. Aquí ves cómo extraer y organizar la información.

````python
# 🎯 EJEMPLO PRÁCTICO: Crear un resumen del clima

def crear_resumen_clima(response):
    """
    Extrae información útil de un JSON complejo de Weather API.
    
    Parámetros:
        response (dict): JSON de la API del clima
        
    Retorna:
        dict: Resumen organizado del clima
    """
    
    # Extraer ubicación (nivel 1)
    ciudad = response['location']['name']
    pais = response['location']['country']
    
    # Extraer clima actual (nivel 1)
    temp_actual = response['current']['temp_c']
    humedad_actual = response['current']['humidity']
    
    # Extraer pronóstico del primer día (nivel 2-3)
    primer_dia = response['forecast']['forecastday'][0]
    fecha = primer_dia['date']
    temp_maxima = primer_dia['day']['maxtemp_c']
    temp_minima = primer_dia['day']['mintemp_c']
    
    # Extraer horas específicas (nivel 4)
    horas = primer_dia['hour']
    
    # Encontrar la hora más calurosa del día
    hora_mas_caliente = max(horas, key=lambda h: h['temp_c'])
    tiempo_max = hora_mas_caliente['time'].split()[1]
    temp_max_hora = hora_mas_caliente['temp_c']
    
    # Encontrar la hora más fría del día
    hora_mas_fria = min(horas, key=lambda h: h['temp_c'])
    tiempo_min = hora_mas_fria['time'].split()[1]
    temp_min_hora = hora_mas_fria['temp_c']
    
    # Crear resumen organizado
    resumen = {
        "ubicacion": f"{ciudad}, {pais}",
        "fecha": fecha,
        "actual": {
            "temperatura": temp_actual,
            "humedad": humedad_actual
        },
        "pronostico_dia": {
            "maxima": temp_maxima,
            "minima": temp_minima
        },
        "analisis_horas": {
            "hora_mas_caliente": {
                "tiempo": tiempo_max,
                "temperatura": temp_max_hora
            },
            "hora_mas_fria": {
                "tiempo": tiempo_min,
                "temperatura": temp_min_hora
            },
            "total_horas": len(horas)
        }
    }
    
    return resumen


# Usar la función
resumen = crear_resumen_clima(response)

# Mostrar el resumen de forma bonita
print("="*60)
print("📍 RESUMEN DEL CLIMA")
print("="*60)
print(json.dumps(resumen, indent=4, ensure_ascii=False))

print("\n" + "="*60)
print("💬 MENSAJE LEGIBLE")
print("="*60)
print(f"El clima en {resumen['ubicacion']} para el {resumen['fecha']}:")
print(f"  • Temperatura actual: {resumen['actual']['temperatura']}°C")
print(f"  • Humedad: {resumen['actual']['humedad']}%")
print(f"  • Máxima del día: {resumen['pronostico_dia']['maxima']}°C")
print(f"  • Mínima del día: {resumen['pronostico_dia']['minima']}°C")
print(f"  • Hora más calurosa: {resumen['analisis_horas']['hora_mas_caliente']['tiempo']} ({resumen['analisis_horas']['hora_mas_caliente']['temperatura']}°C)")
print(f"  • Hora más fría: {resumen['analisis_horas']['hora_mas_fria']['tiempo']} ({resumen['analisis_horas']['hora_mas_fria']['temperatura']}°C)")
````


## 🎓 Resumen Final

**JSON tiene solo 2 estructuras:**
1. **Objetos `{}`** → Acceso por clave: `data['clave']`
2. **Arrays `[]`** → Acceso por índice: `data[0]`

**Para dominar JSON:**
1. Identifica si es objeto `{}` o array `[]`
2. Baja nivel por nivel
3. Usa `.keys()`, `type()`, `len()` para explorar
4. Imprime con `json.dumps(data, indent=4)` para ver la estructura5. Siempre valida con `.get()` cuando no estés seguro
